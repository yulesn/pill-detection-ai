```python
from pathlib import Path
from collections import Counter

PROJECT_ROOT = Path(__file__).resolve().parents[1]

TRAIN_LABEL_DIR = (
    PROJECT_ROOT
    / "data"
    / "combined_subset"
    / "labels"
    / "train"
)

NUM_CLASSES = 56


counts = Counter()

aihub_counts = Counter()
original_counts = Counter()

for label_file in TRAIN_LABEL_DIR.glob("*.txt"):

    is_aihub = label_file.name.startswith("aihub_")

    with open(label_file, "r", encoding="utf-8") as f:

        for line in f:

            parts = line.strip().split()

            if len(parts) != 5:
                continue

            try:
                class_id = int(parts[0])
            except ValueError:
                continue

            if not 0 <= class_id < NUM_CLASSES:
                continue

            counts[class_id] += 1

            if is_aihub:
                aihub_counts[class_id] += 1
            else:
                original_counts[class_id] += 1


print("=" * 70)
print("Combined Train 실제 클래스별 객체 수")
print("=" * 70)

print(
    f"{'Class':<10}"
    f"{'기존':>10}"
    f"{'AI-Hub':>10}"
    f"{'최종':>10}"
)

print("-" * 70)

for class_id in range(NUM_CLASSES):

    print(
        f"class_{class_id:<4}"
        f"{original_counts[class_id]:>10}"
        f"{aihub_counts[class_id]:>10}"
        f"{counts[class_id]:>10}"
    )


print()
print("=" * 70)
print("통계")
print("=" * 70)

print(f"기존 객체 수  : {sum(original_counts.values()):,}")
print(f"AI-Hub 객체 수: {sum(aihub_counts.values()):,}")
print(f"최종 객체 수  : {sum(counts.values()):,}")


print()
print("=" * 70)
print("AI-Hub 추가 객체 TOP 10")
print("=" * 70)

for class_id, count in aihub_counts.most_common(10):
    print(f"class_{class_id:02d} : {count:4d}개")


print()
print("=" * 70)
print("AI-Hub 추가 객체 BOTTOM 10")
print("=" * 70)

for class_id in sorted(
    range(NUM_CLASSES),
    key=lambda x: aihub_counts[x]
):

    print(
        f"class_{class_id:02d} : "
        f"{aihub_counts[class_id]:4d}개"
    )
```
